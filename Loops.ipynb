# **While Loop**

```python
# print values 1 to 20
```

```python
c=1 # initialization
```

```python
while c<=20: #condition
  print(c)
  c=c+1 #incremental
```

```python
#print values from 20 to 1 using while loop
a=20
```

```python
while a>=1: #condition
  print(a)
  a=a-1 #decremental
```

# **Loop with conditional statements**

```python
#print the values from 1 to 10 without 5
#1,2,3,4,6,7,8,9,10
```

```python
for i in range(1,11):
# print(i)
  if i==5:
    continue
  print(i)
```

```python
for i in range(1,11):
# print(i)
  if i==5:
    continue
  else:
    print(i)
```

```python
for i in range(1,11):
# print(i)
  if i==5:
    break
  else:
    print(i)
```

```python
#print 0,1,3,4
for i in range(0,5):
  if i==2:
    continue
  else:
    print(i)
```
