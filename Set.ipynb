# **SET operations**

```python
a={1,2,3,4}
b={3,4,5,6}
```

```python
a,b
```

```python
print(a)
print(b)
```

```python
# Union
a|b
```

```python
a.union(b)
```

```python
# Intersection
a&b
```

```python
a.intersection(b)
```

```python
# Differnce
a-b
```

```python
a.difference(b)
```

```python
b-a
```

```python
b.difference(a)
```

```python
# Symmetric difference
a^b
```

```python
a.symmetric_difference(b)
```
