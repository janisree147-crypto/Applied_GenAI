**create a ranking system - A(90-100), B(75-89), C(60-74), D(50-59), E(35-49), F(0-34)**

```python
score=55
```

```python
if score>=90:
  print("A")
elif score>=75 and score<90:
  print("B")
elif score>=60 and score<75:
  print("C")
elif score>=50 and score<60:
  print("D")
elif score>=35 and score<50:
  print("E")
else:
  print("F")
```

```python
if score>=90:
  print("A")
elif score>=75 :
  print("B")
elif score>=60 :
  print("C")
elif score>=50 :
  print("D")
elif score>=35 :
  print("E")
else:
  print("F")
```
